```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
#Parameters
```

```python
T0 = 5   # initial condition
k = 0.2
T_eq = 20
```

```python
#Time Range
```

```python
t = np.linspace(0, 50, 200)
```

```python
#Solution
```

```python
T = T_eq + (T0 - T_eq) * np.exp(-k * t)
```

```python
#plot
```

```python
# Plot
plt.figure(figsize=(8,5))
plt.plot(t, T, label=f"T(t), T0={T0}")
plt.axhline(T_eq, color='red', linestyle='--', label="Equilibrium T=20")
plt.xlabel("Time t")
plt.ylabel("Temperature T")
plt.title("Solution of dT/dt = 0.2(20 - T)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#draw it as a linear plot 
```

```python
# Define Temperature range
T = np.linspace(0, 30, 100)

# Derivative function
dT_dt = 0.2 * (20 - T)

# Plot
plt.figure(figsize=(8, 5))
plt.plot(T, dT_dt, color='blue', linewidth=2)

# Labels and title
plt.title("Derivative: 0.2(20 - T)")
plt.xlabel("Temperature (T)")
plt.ylabel("Rate of Change (dT/dt)")

# Grid and reference line
plt.axhline(0, color='black', linewidth=1)
plt.grid(True, linestyle='--', alpha=0.6)
```

```python
#Numerical Methods
```

```python
# Parameters
dt = 0.1   # time step
steps = 100
initial_conditions = [5, 10, 25, 30]  # different starting temperatures

# Simulation
for T0 in initial_conditions:
    T = [T0]
    for _ in range(steps):
        dT_dt = 0.2 * (20 - T[-1])
        T.append(T[-1] + dt * dT_dt)
    plt.plot(np.arange(steps+1)*dt, T, label=f"T0={T0}")

# Plot
plt.axhline(20, color='red', linestyle='--', label="Fixed Point T=20")
plt.xlabel("Time")
plt.ylabel("Temperature")
plt.title("Numerical Simulation of dT/dt = 0.2(20 - T)")
plt.legend()
plt.grid(True)
```
